```python
def get_business_kpis():
    """Return the main business KPIs."""

    total_orders = int(df_clean['OrderID'].nunique())
    total_quantity = int(df_clean['Quantity'].sum())
    total_revenue = float(df_clean['TotalAmount'].sum())
    average_order_value = float(total_revenue / total_orders)

    return {
        "total_revenue": total_revenue,
        "total_orders": total_orders,
        "total_quantity": total_quantity,
        "average_order_value": average_order_value
    }
```

```python
# ==========================================
# LOAD ALL AI ANALYST TOOLS
# ==========================================

def get_top_products():
    """Return top 10 products by revenue."""

    result = (
        df_clean
        .groupby('ProductName')['TotalAmount']
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    return {
        str(product): float(revenue)
        for product, revenue in result.items()
    }


def get_revenue_by_brand():
    """Return revenue by brand."""

    result = (
        df_clean
        .groupby('Brand')['TotalAmount']
        .sum()
        .sort_values(ascending=False)
    )

    return {
        str(brand): float(revenue)
        for brand, revenue in result.items()
    }


def get_revenue_by_category():
    """Return revenue by category."""

    result = (
        df_clean
        .groupby('Category')['TotalAmount']
        .sum()
        .sort_values(ascending=False)
    )

    return {
        str(category): float(revenue)
        for category, revenue in result.items()
    }


def get_monthly_revenue():
    """Return monthly revenue."""

    result = (
        df_clean
        .groupby(df_clean['OrderDate'].dt.to_period('M'))['TotalAmount']
        .sum()
        .sort_values(ascending=False)
    )

    return {
        str(month): float(revenue)
        for month, revenue in result.items()
    }


def get_order_status():
    """Return order counts by status."""

    result = (
        df_clean
        .groupby('OrderStatus')['OrderID']
        .nunique()
        .sort_values(ascending=False)
    )

    return {
        str(status): int(count)
        for status, count in result.items()
    }


def get_payment_method_revenue():
    """Return revenue by payment method."""

    result = (
        df_clean
        .groupby('PaymentMethod')['TotalAmount']
        .sum()
        .sort_values(ascending=False)
    )

    return {
        str(method): float(revenue)
        for method, revenue in result.items()
    }


def get_city_revenue():
    """Return revenue by city."""

    result = (
        df_clean
        .groupby('City')['TotalAmount']
        .sum()
        .sort_values(ascending=False)
    )

    return {
        str(city): float(revenue)
        for city, revenue in result.items()
    }


def get_state_revenue():
    """Return revenue by state."""

    result = (
        df_clean
        .groupby('State')['TotalAmount']
        .sum()
        .sort_values(ascending=False)
    )

    return {
        str(state): float(revenue)
        for state, revenue in result.items()
    }


print("All analytical tools loaded successfully!")
```

```python
# ==========================================
# AUTOMATIC KPI VALIDATION
# ==========================================

def validate_kpis():

    # Actual values from cleaned dataset
    actual_orders = df_clean["OrderID"].nunique()
    actual_quantity = df_clean["Quantity"].sum()
    actual_revenue = df_clean["TotalAmount"].sum()
    actual_aov = actual_revenue / actual_orders

    # Expected values from AI/tool result
    tool_result = get_business_kpis()

    # Compare values
    orders_match = actual_orders == tool_result["total_orders"]
    quantity_match = actual_quantity == tool_result["total_quantity"]
    revenue_match = round(actual_revenue, 2) == round(tool_result["total_revenue"], 2)
    aov_match = round(actual_aov, 2) == round(tool_result["average_order_value"], 2)

    verified = all([
        orders_match,
        quantity_match,
        revenue_match,
        aov_match
    ])

    return {
        "Total Orders": orders_match,
        "Total Quantity": quantity_match,
        "Total Revenue": revenue_match,
        "Average Order Value": aov_match,
        "Overall Status": "VERIFIED" if verified else "NOT VERIFIED"
    }


validation = validate_kpis()

print("=== AUTOMATIC VALIDATION ===")

for metric, status in validation.items():
    print(f"{metric}: {status}")
```

```python
# ==========================================
# VERIFIED AI BUSINESS RESPONSE
# ==========================================

def get_verified_kpi_response():

    # Get the actual validated data
    validation = validate_kpis()

    if validation["Overall Status"] == "VERIFIED":
        verification_message = "✓ VERIFIED — All numerical results match the dataset."
    else:
        verification_message = "⚠ NOT VERIFIED — Results require review."

    return verification_message


print(get_verified_kpi_response())
```

```python
# ==========================================
# BUSINESS INSIGHT - TOP REVENUE CATEGORY
# ==========================================

def generate_category_insight():

    category_revenue = (
        df_clean.groupby("Category")["TotalAmount"]
        .sum()
        .sort_values(ascending=False)
    )

    top_category = category_revenue.index[0]
    top_revenue = category_revenue.iloc[0]

    total_revenue = df_clean["TotalAmount"].sum()

    revenue_share = (top_revenue / total_revenue) * 100

    print("=== BUSINESS INSIGHT ===")
    print(f"Top Revenue Category: {top_category}")
    print(f"Revenue: ${top_revenue:,.2f}")
    print(f"Share of Total Revenue: {revenue_share:.2f}%")

    print("\n=== RECOMMENDATION ===")
    print(
        f"Focus on {top_category} by investigating its "
        "top-performing products and locations. "
        "Prioritize inventory and marketing for products "
        "that consistently generate strong revenue."
    )


generate_category_insight()
```

```python
# ==========================================
# TOP PRODUCTS IN TOP REVENUE CATEGORY
# ==========================================

def get_top_products_in_category(category, n=5):

    result = (
        df_clean[df_clean["Category"] == category]
        .groupby("ProductName")["TotalAmount"]
        .sum()
        .sort_values(ascending=False)
        .head(n)
    )

    return result


top_category = (
    df_clean.groupby("Category")["TotalAmount"]
    .sum()
    .idxmax()
)

top_products = get_top_products_in_category(top_category)

print(f"=== TOP PRODUCTS IN {top_category.upper()} ===")

for product, revenue in top_products.items():
    print(f"{product}: ${revenue:,.2f}")
```
